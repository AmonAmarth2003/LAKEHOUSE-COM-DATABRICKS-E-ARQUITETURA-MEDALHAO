## df = spark.read.option("infeschema", "true").option("header", "true").csv("path + filename")

Gera um dataframe para cada arquivo que está no Volume "dados".


```python
caminho_landing = '/Volumes/workspace/landing/dados'

df_agent_policies     = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/agent_policies.csv")
df_agents             = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/agents.csv")
df_claims             = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/claims.csv")
df_customers          = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/customers.csv")
df_insurance_types    = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/insurance_types.csv")
df_payments           = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/payments.csv")
df_policies           = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/policies.csv")
```

## df = df_apolice.withColumn("nome_coluna", "valor")

Adiciona uma nova coluna (metadado) de data e hora de processamento e nome do arquivo de origem.

```python
from pyspark.sql.functions import current_timestamp, lit

df_agent_policies = df_agent_policies.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("agent_policies.csv"))
df_agents = df_agents.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("agents.csv"))
df_claims = df_claims.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("claims.csv"))
df_customers = df_customers.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("customers.csv"))
df_insurance_types = df_insurance_types.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("insurance_types.csv"))
df_payments = df_payments.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("payments.csv"))
df_policies = df_policies.withColumn("date_hour_bronze", current_timestamp()).withColumn("file_name", lit("policies.csv"))
```

## df_apolice.write.format('delta').saveAsTable("nome_tabela")

Salva os dataframes em arquivos delta lake (formato de arquivo) no schema/database "bronze". As tabelas geradas são do tipo MANAGED (gerenciadas).

```python
df_agent_policies.write.format('delta').mode("overwrite").saveAsTable("bronze.agent_policies")
df_agents.write.format('delta').mode("overwrite").saveAsTable("bronze.agents")
df_claims.write.format('delta').mode("overwrite").saveAsTable("bronze.claims")
df_customers.write.format('delta').mode("overwrite").saveAsTable("bronze.customers")
df_insurance_types.write.format('delta').mode("overwrite").saveAsTable("bronze.insurance_types")
df_payments.write.format('delta').mode("overwrite").saveAsTable("bronze.payments")
df_policies.write.format('delta').mode("overwrite").saveAsTable("bronze.policies")
```
